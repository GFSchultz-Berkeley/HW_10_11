## 3. Answering the questions (20 points)

Now that the data is cleaned and filterd - let's answer the two questions from your boss! That is use the dataframe from 2e above that has done all of the cleaning & filtering steps!

**3a.** Which zipcode had the highest count of contributions and the most dollar amount? (10 points)

```python
# 3a YOUR CODE HERE
by_zip = df.groupby('zip5')['contb_receipt_amt'].agg(['count', 'sum'])

print('Most contributions:')
print(by_zip.sort_values('count', ascending=False).head())
print('\nMost dollars:')
print(by_zip.sort_values('sum', ascending=False).head())

print('\nTop ZIP by count:', by_zip['count'].idxmax(), '| Top ZIP by dollars:', by_zip['sum'].idxmax())
```

- **3a answer here:** ZIP code 94110, which is the Mission district of San Fransisco had both the most contributions (3,799) and the most total dollars ($284,398.05)

**3b.** What day(s) of the month do most people donate? (10 points)

```python
# 3b YOUR CODE HERE
day = df['contb_receipt_dt'].dt.day

by_day = day.value_counts().sort_index()
print(by_day)
print('\nTop 5 days:')
print(by_day.sort_values(ascending=False).head())

days_in_range = pd.Series(pd.date_range(df['contb_receipt_dt'].min(), df['contb_receipt_dt'].max()))
per_day = by_day / days_in_range.dt.day.value_counts()
print('\nTop 5 days, per occurrence:')
print(per_day.sort_values(ascending=False).head())

print('\nBusiest dates:')
print(df['contb_receipt_dt'].value_counts().head(8))
```

- **3b answer here:** Most people tend to donate at the end of the month, with the busiest days being the 29th, 30th, and 31st. This makes sense as campaigns adjust their email lists to respond to end of month asks, and giving builds through the last week with a big spike on the last day.
